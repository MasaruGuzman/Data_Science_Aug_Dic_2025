## Filtering and Subsetting Data

In order to protect the identity of the passengers, you will delete the `Name` column. You can remove rows or columns from the dataframe using the method `pd.drop()`. The argument of this method is a list with the following characteristics:

- If you want to remove rows, `pd.drop([row1, row2, ... ])` will remove the rows with the given indices.
- If you want to remove columns, you can use either `pd.drop([col1, col2, ... ], axis = 1)` or `pd.drop(columns = [col1, col2, ... ])`

Don't forget to use the argument `inplace = True` to actually drop the column.

```python
# Write here the code for removing the 'Name' column from the dataframe, and then display column names to verify that the column was dropped
titanic_data.drop(columns=['Name'], inplace=True)
titanic_data.columns
```

## Conditional filtering

Conditional filtering is one of the most important parts of data exploration as it helps you uncover different patterns in data based on filters. For conditional filtering, you are provided with different comparison operations like `==`, `>`, `<`, and `!=`, along with the logical operators like `and`, `or` and `not`. The basic expression of filtering requires selecting a column and a filter value or a list of filter values plugged into `[]` brackets. For example, if we want to filter the `Pclass` (passenger class) in the Titanic dataset, we write

```python
titanic_data[titanic_data['Pclass'] == 3]
```

You can also apply multiple conditions to a single or a set of columns using the logical operators. Filter the Titanic data where `Pclass` is `3` and `Age` of passengers is greater than or equal to `30`.

```python
# Write here the code to filter the Titanic data where `Pclass` is `3` and `Age` of passengers is greater than or equal to `30`.
titanic_data[(titanic_data['Pclass'] == 3) & (titanic_data['Age'] >= 30)]
```

## Basic Data Exploration

By now, you’ll start to understand your data and its different attributes. You can now go deeper into the exploration and try out a few concepts like value counts, grouping, and aggregations.

### Value Counts

Checking the frequency of values in a column is a very common practice used for data analysis. When working on machine learning use cases, sometimes your data may be imbalanced, and the `pd.value_counts()` method help you determine the ratio of target classes. It provides the count of unique values in a column. Display value counts for the `Sex` column.

```python
# Write here the code for displaying value counts of the `Sex` column
titanic_data.value_counts('Sex')
```

- How many male passengers are listed in the data frame? <u>266</u>
- How many female passengers are listed in the data frame? <u>152</u>

You can also calculate the multi-index value counts in the dataset by slicing the data frame with a list of the desired columns. Display the value counts of the columns `Pclass` and `Age`.

```python
# Write here the code to display the value counts of the columns Pclass and Age
titanic_data.value_counts(subset=['Pclass', 'Age'])
```

- How many 21 year old third class passengers are listed? <u>11</u>
- How many 30 year old second class passengers are listed? <u>7</u>

### Grouping and Aggregation

Sometimes you might also need to group the data based on different statistics such as mean, sum, and median that reveal a lot of hidden patterns in the data. Pandas `pd.groupby()` method helps you group the larger set of data and allows you to perform different operations on it. To use the `pd.groupby()` method, you need to provide the feature name or a list of feature names and an aggregate function based on which the data will be grouped together. For example, if you want to group the Titanic data based on the `Sex` column and with `mean()` aggregation, you can write `titanic_data.groupby(['Sex']).mean(numeric_only = True)`. Note that it is necessary to indicate that you'll work with numeric only columns. Try it.

```python
## Write here to calculate the mean of all numerical columns grouped by Sex
titanic_data.groupby(['Sex']).mean(numeric_only = True)
```

With `pd.groupby()`, we can see that the mean value of women surviving the Titanic crash is 1. This is true, as most women on board survived the incident (though shows the limitations of mean as a central statistic). Grouping data can also be multi-index where you can group data based on two or more categories. Display a table with the mean of all numerical columns in the data frame grouped by `Sex` and `Pclass`.

```python
# Write here the code to display a table with the mean of all numerical columns in the data frame grouped by Sex and Pclass
titanic_data.groupby(['Sex', 'Pclass']).mean(numeric_only = True)
```

- What is the average age for second class female passengers? <u>24.376552</u>
- What is the average fare paid by first class male passengers? <u>75.586551</u>

## Data Cleaning and Preprocessing

Now come to the most important part of exploratory data analysis that helps shape your data for ML analysis. The real-world data is not always ideal and can not be used directly for ML analysis as it can contain NaN values, inaccurate data, duplicate values, or corrupted values. Pandas provides a lot of built-in functions to deal with these values.

### Checking and Handling Missing Values

The majority of machine learning and deep learning models can not deal with NaN values, so you need to handle them before passing your data to these models. Three normal approaches to handling these NaN values are:

- Predicting the NaN values based on the other features in the dataset.
- Dropping the NaN values from the dataset (not often recommended).
- Replacing the NaN values with the most appropriate values (often with the measure of central tendency).

### Dropna

**pandas** provides the `pd.dropna()` method to drop null values based on rows and columns. You might have noticed above that the `Age`, `Fare`, and `Cabin` columns contain the NaN values. Create a new data frame named `titanic_data_dropna` buy using the `pd.dropna()` method with the argument `axis=1` (this removes columns with missing data) and display the `info()` of the new data frame.

```python
# Write here the code to create a new data frame named titanic_data_dropna buy using the pd.dropna() method with the argument axis=1 and display its info
titanic_datadropna = titanic_data.dropna(axis = 1)
titanic_datadropna.info()
```

- Were the columns `Age`, `Fare`, and `Cabin` removed from the data frame? <u>Yes</u>

### Fillna

Another method provided by Pandas is `pd.fillna()` that replaces the NaN values with the provided custom values. For example, if you want to replace the NaN values in the `Age` column with an average age in the dataset, you can do so as follows:

```python
# replace NaN values
titanic_data_fillna_age = titanic_data.copy()
titanic_data_fillna_age.fillna({'Age': titanic_data_fillna_age['Age'].mean()},inplace=True)
```

We can display the results of the above by using the `pd.info()` or `pd.describe()` methods:

```python
titanic_data_fillna_age['Age'].describe()
```

Now it's your turn. Fill the missing values of the `Fare` column using the above example as a template. Use the same name as above when creating a copy of the original data frame.

```python
# Write here the code for filling the missing values of the Fare column using the above example as a template.
# Use a different name when creating a copy of the original data frame. Display the description of the new data frame
# replace NaN values
titanic_data_fillna_fare = titanic_data.copy()
titanic_data_fillna_fare.fillna({'Fare': titanic_data_fillna_fare['Fare'].mean()},inplace=True)
titanic_data_fillna_fare['Fare'].describe()
```

### Convert Data Types of columns

Sometimes saving the data in files like CSV can lose the original datatypes. For example, a `float` parameter could be saved as `int`. You can change these datatypes with the help of `pd.astype()` method. Convert the `float` values of the `Age` column of the `titanic_data_fillna_age` data frame to `int`. You can display the `info()` of the data frame to verify the change.

```python
# Write here the code to convert the float values of the Age column of the titanic_data_fillna_age data frame to int
# and verify that the change was made

titanic_data_fillna_age['Age'] = titanic_data_fillna_age['Age'].astype(int)
titanic_data_fillna_age.info()
```
