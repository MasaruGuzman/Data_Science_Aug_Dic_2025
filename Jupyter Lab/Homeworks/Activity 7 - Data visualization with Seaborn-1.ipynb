## Scatterplots

Draw a scatter plot of `bill_depth_mm` vs `bill_length_mm`, colored by `species`.

```python
# Write here the code to generate a scatter plot of bill_depth_mm vs bill_length_mm
sns.scatterplot(x = "bill_length_mm", y = "bill_depth_mm", hue = "species", data = data)
```

## Histograms and density plots

Create a histogram for `flipper_length_mm`.

```python
# Write here your code for creating a histogram for flipper_length_mm
sns.histplot(data = data, x = "flipper_length_mm")
plt.xlabel("Flipper Length (mm)")
```

Create a histogram of `body_mass_g' with horizontal bars.

```python
# Write here your code for creating a histogram of `body_mass_g' with horizontal bars
sns.histplot(data = data, y = "body_mass_g")
plt.ylabel("Body Mass (g)")
```

Create a histogram of `bill_depth_mm` with the distribution shape on top.

```python
# Write here your code for creating a histogram of `bill_depth_mm` with the distribution shape on top
sns.histplot(data = data, x = "bill_depth_mm", kde = True)
plt.xlabel("Bill depth (mm)")
```

Create a histogram of `flipper_length_mm` colored by `species`.

```python
# Write here your code for creating a histogram of `flipper_length_mm` colored by `species`
sns.histplot(data = data, x = "flipper_length_mm", hue = "species")
```

## Barplots

Create a barplot of `flipper_length_mm` separated and colored by `species`.

```python
# Write here your code for creating a barplot of `flipper_length_mm` separated and colored  by `species`
sns.barplot(data = data, x = "species", y = "flipper_length_mm", hue = "species")
plt.xlabel("Penguin Species")
plt.ylabel("Avg Flipper Length (mm)")
```

Create a barplot of `flipper_length_mm` separated by `species` and colored by `sex`.

```python
# Write here your code for creating a barplot of `flipper_length_mm` separated by `species` and colored by `sex`
sns.barplot(data = data, x = "species", y = "flipper_length_mm", hue = "sex")
plt.xlabel("Penguin Species")
plt.ylabel("Avg Flipper Length (mm)")
```

## Boxplots and violin plots

Create boxplots of `bill_length_mm` separated and colored by `species`.

```python
# Write here your code for creating boxplots of `bill_length_mm` separated and colored by `species`
sns.boxplot(data = data,x = "species", y = "bill_length_mm", hue = "species")
plt.xlabel("Penguin species")
plt.ylabel("Bill length (mm)")
```

Create boxplots of `bill_depth_mm` separated by `species` and colored by `sex`.

```python
# Write here your code for creating boxplots of `bill_depth_mm` separated by `species` and colored by `sex`
sns.boxplot(data = data,x = "species", y = "bill_length_mm", hue = "sex")
plt.xlabel("Penguin species")
plt.ylabel("Bill length (mm)")
```

Now, reproduce the previous two graphs but using violin plots instead of boxplots.

```python
# Write here your code for creating violin plots of `bill_length_mm` separated and colored by `species`
sns.violinplot(data = data,x = "species", y = "bill_length_mm", hue = "species")
plt.xlabel("Penguin species")
plt.ylabel("Bill length (mm)")
```

```python
# Write here your code for creating violin plots of `bill_depth_mm` separated by `species` and colored by `sex`
sns.violinplot(data = data,x = "species", y = "bill_length_mm", hue = "sex")
plt.xlabel("Penguin species")
plt.ylabel("Bill length (mm)")
```

## Facet grid plots

Create a facet grid plot of histograms for `flipper_length_mm` with `island` as columns and `sex` as rows.

```python
# Write here your code for creating a facet grid plot of histograms for `flipper_length_mm` with `island` as columns and `sex` as rows
g = sns.FacetGrid(data, col= "island", row = "sex")
g.map(sns.histplot, "flipper_length_mm")
```

Create a facet grid plot of filled in distribution plots for `flipper_length_mm` with `island` as columns and `sex` as rows.

```python
# Write here your code for creating a facet grid plot of filled in distribution plots for `flipper_length_mm` with `island` as columns and `sex` as rows
g = sns.FacetGrid(data, col = "island", row = "sex")
g.map(sns.kdeplot, "flipper_length_mm")
```

## Paired plots

Create a paired plot for your data set with distribution plots on the diagonal, colored by `species`.

```python
# Write here your code for creating a paired plot for your data set with distribution plots on the diagonal, colored by `species`
sns.pairplot(data = data, hue = "species", diag_kind = "kde", palette = "pastel")
```

Recreate the previous paired plot, but with histograms on the diagonal.

```python
# Write here your code for creating a paired plot for your data set with histograms on the diagonal, colored by `species`
sns.pairplot(data = data, hue = "species", diag_kind = "hist", palette = "pastel")
```

## Heatmaps

Create a heatmap of the correlations of the variables in your data set with annotated cells. Use `data.corr(numeric_only=True)` for obtaining correlations.

```python
# Write here your code for creating a heatmap of the correlations of the variables in your data set
corr = data.corr(numeric_only = True)
sns.heatmap(corr, annot = True, cmap = "coolwarm")
plt.title("Correlation Heatmap")
```
