## Customizing a plot

There are several `plt` arguments that allow you to customize how the data is presented in a plot

- `marker='o'` → adds data markers to a plot. You can check other option [here](https://matplotlib.org/stable/api/markers_api.html).
- `linestyle='--'` → changes the line style. [Options for line style](https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html). The option`'None'` omits the line and leaves the markers.
- `color='b'` → changes the color of the data. [Color options](https://matplotlib.org/stable/tutorials/colors/colors.html)
- `label='name'` → cross-reference label for the plot.

Furthermore, the object `Axes` includes several methods that can be used to identify the different parts of the plot. Set these options before `plt.show()`.

- `.set_xlabel('Name')` → name for the X axis.
- `.set_ylabel('Name')` → name for the Y axis.
- `.set_title('Name')` → plot title.
- `.set(xlabel='Name')` → alternative to `set_xlabel('Name')`. The other options have similar alteranitve versions.
- `.legend()` → tags for the differen graphs in the plot.

```python
fig, ax = plt.subplots()
ax.plot(messi_goals.year, messi_goals.goals,
        marker='o', color='b', linestyle='--', label='Lionel Messi')
ax.plot(ronaldo_goals.year, ronaldo_goals.goals,
        marker='s', color='r', linestyle='dotted', label='Cristiano Ronaldo')

ax.set(xlabel='Years', ylabel='Goals', title='Messi vs. Cristiano goals (2014 - 2019)')
ax.legend()
plt.savefig('2.jpg')
plt.savefig('2.pdf')
plt.show()
```

## Bar chart

```python
# Red cards per team
top_red_cards = df.groupby('team_name')['red_cards'].sum() \
                .sort_values(ascending=False).head(10)
top_red_cards
```

```python
# Plot
# plt.figure(figsize=[6.4, 6])
plt.style.use('ggplot')
plt.bar(top_red_cards.index, top_red_cards)
plt.xticks(rotation=60)
plt.title('Cumulative red cards per team')
plt.ylabel('Red cards')
plt.show()
```

```python
plt.barh(top_red_cards.index, top_red_cards)
plt.title('Cumulative red cards per team')
plt.xlabel('Red cards')
plt.show()
```

### Stacked bars chart

```python
# Filter
df['goals_assists'] = df.goals + df.assists
top_players = df.groupby('player_name')[['goals', 'assists', 'goals_assists']].sum() \
                .sort_values('goals_assists', ascending=False).head(10)
top_players
```

```python
# Plot
plt.bar(top_players.index, top_players.goals, label='Goals')
plt.bar(top_players.index, top_players.assists, label='Assists', bottom=top_players.goals)
plt.xticks(rotation=90)
plt.title('Goals and assists for top players')
plt.ylabel('Goals and assists')
plt.legend()
plt.show()
```

## Histograms

Use `plt.hist(X)` for drawing histograms. The argument is the column to be analized. It can be customized with the following options:

- `bins=n` → how many bars will be used. Default = 10.
- It's possible to include a list with the range of values for X that we want to display
- If we want to show two histograms on the same plot, we use the option `alpha`, which is a number between 0 and 1, and indicates the transparency of one of the histograms.
- The option `histtype='step'` changes the visual presentation of the histogram. Other options can be found [here](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html).
- The option `density=True` displays values as percentages.

```python
# Goles totales por equipo en 2018
df_2018 = df.query('year == 2018')
team_goals_2018 = df_2018.groupby('team_name')['goals'].sum()

# Plot
plt.figure(figsize=[7, 6]) # 7 in de ancho y 6 in de alto
plt.hist(team_goals_2018, bins=15,color='b')
plt.title('Total goals per team in 2018')
plt.xlabel('Number of goals')
plt.ylabel('Number of teams with X number of goals')
#plt.savefig('6.jpg')
plt.show()
```

```python
df_2018 = df.query('year == 2018')
team_goals_2018 = df_2018.groupby('team_name')['goals'].sum()
plt.hist(team_goals_2018, bins=15, histtype='step', density=True)
plt.show()
```

## Sharing Matplotlib plots

- Use `fig.savefig('path/name.png')` to save the figure on disk. File format is determined with the extension given to the file
- The argument `quality`, which takes values between `1` and `100`, allows to change the quality of the image.
- `dpi=300` saves the file with a resolution of 300dpi. You may change this value.
- Figure size can be set with `fig.set_size_inches([width, height])`.

```python
fig.set_size_inches([5, 3])
fig.savefig('figure_5_3.png')
```
