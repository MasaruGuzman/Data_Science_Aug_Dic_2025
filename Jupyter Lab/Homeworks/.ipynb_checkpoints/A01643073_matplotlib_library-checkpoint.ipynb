## Activity 6 - Data visualization with Matplotlib

### 1.- Load libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### 2.- Load the dataset

This [dataset](https://www.kaggle.com/datasets/spscientist/students-performance-in-exams) can be found on Kaggle and it is also provided on Canvas.

```python
df_students_performance = pd.read_csv('StudentsPerformance.csv')
```

### 3.- Set up your variables

For your plots you'll use three columns from the dataset. Extract the variables `math_score`, `reading_score` and `writing_score` as a `Pandas Series`.

```python
math_score = df_students_performance['math score']
reading_score = df_students_performance['reading score']
writing_score = df_students_performance['writing score']
```

This way you don't have to slice the data frame everytime you need to use them.

### 4.- Complete the following exercises

a) Create a histogram for each of these variables with 10 bins: `math_score`, `reading_score`, `writing_score`.

b) Create a boxplot for each of these variables: `math_score`, `reading_score`, `writing_score`. Remember to customize your boxplots so they don't look all the same.

c) Use **matplolib** grammar to create a visualiztion with 6 axes (3, 2). On the left column plot the histograms, and on the right plot the boxplots.

d) Plot 3 scatter plots to see the relationship among the 3 variables. Apply an opacity of `alpha = 0.5`:
- `math_score` - `reading_score`
- `math_score` - `writing_score`
- `reading_score` - `writing_score`

In order to modify the color of the points, apply the following:

```python
N = 1000
colors = np.random.rand(N)
```

This will give you 1000 color possibilities for your scatter plots using the color parameter `c = colors`.

e) Apply color and style modifications to all the graphs you created. Don't forget to include titles for the graphs and for the `x` and `y` axes.

```python
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(10, 12))

# Histograms
axes[0, 0].hist(math_score, bins=10, edgecolor='blue', color='lightblue')
axes[0, 0].set(title='Math Score', ylabel='# of people')
axes[0, 0].grid(True)

axes[1, 0].hist(reading_score, bins=10, edgecolor='red', color='yellow')
axes[1, 0].set(title='Reading Score', ylabel='# of people')

axes[2, 0].hist(writing_score, bins=10, edgecolor='green', color='lightgreen')
axes[2, 0].set(title='Writing Score', xlabel='Grade', ylabel='# of people')
axes[2, 0].grid(True)

# Boxplots
axes[0, 1].boxplot(math_score, vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue', edgecolor='blue'))
axes[0, 1].set(title='Math Score')

axes[1, 1].boxplot(reading_score, vert=False, patch_artist=True, boxprops=dict(facecolor='yellow', edgecolor='red'))
axes[1, 1].set(title='Reading Score')
axes[1, 1].grid(True)

axes[2, 1].boxplot(writing_score, vert=False, patch_artist=True, boxprops=dict(facecolor='lightgreen', edgecolor='green'))
axes[2, 1].set(title='Writing Score')

# Scatters
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(18, 5))

axes[0].scatter(math_score, reading_score, alpha=0.5, c=colors)
axes[0].set(title='Math vs. Reading', xlabel='Math Score', ylabel='Reading Score')
axes[0].grid(True)
    
axes[1].scatter(math_score, writing_score, alpha=0.5, c=colors)
axes[1].set(title='Math vs. Writing', xlabel='Math Score', ylabel='Writing Score')

axes[2].scatter(reading_score, writing_score, alpha=0.5, c=colors)
axes[2].set(title='Reading vs. Writing', xlabel='Reading Score', ylabel='Writing Score')
axes[2].grid(True)



plt.show()
```
