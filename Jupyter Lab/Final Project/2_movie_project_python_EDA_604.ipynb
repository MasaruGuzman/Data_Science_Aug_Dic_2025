#### Detecting outliers

In order to detect outliers, we create boxplots for all numerical variables.

```python
# Remember to change the colors of the bloxplots

fig, axes = plt.subplots(3, 3,figsize=(10,10))
fig.tight_layout(pad=3)
sns.boxplot(x=movies['Runtime'], ax=axes[0,0], color='skyblue');
sns.boxplot(x=movies['Budget'], ax=axes[0,1], color='skyblue');
sns.boxplot(x=movies['Revenue'], ax=axes[0,2], color='skyblue');

sns.boxplot(x=movies['IMDb'], ax=axes[1,0], color='lightgreen');
sns.boxplot(x=movies['Metacritic'], ax=axes[1,1], color='lightgreen');
sns.boxplot(x=movies['Rotten_Tomatoes'], ax=axes[1,2], color='lightgreen');

sns.boxplot(x=movies['Filmaffinity'], ax=axes[2,0], color = "salmon");
sns.boxplot(x=movies['Personal'], ax=axes[2,1], color = "salmon");
sns.boxplot(x=movies['Fans'], ax=axes[2,2], color = "salmon");
```

Our approach for removing outliers will be naive: We calculate the interquartile range, `IQR`, for all numerical variables, and remove all records more than `1.5*IQR` units from the nearest quartile. This reduces the interquartile range and will possible "create" more outliers, but they will be within the endpoints of the whiskers of the previous plots.

```python
movies_numeric = movies.select_dtypes(include='number')
```

```python
Q1 = movies_numeric.quantile(0.25)
Q3 = movies_numeric.quantile(0.75)
IQR = Q3 - Q1
IQR
```

```python
movies_outliers=movies[((movies_numeric < (Q1 - 1.5 * IQR)) |(movies_numeric > (Q3 + 1.5 * IQR))).any(axis=1)]
movies_outliers
```

```python
movies.drop(movies_outliers.index,inplace=True)
movies.shape
```

We are left with 167 records, still enough for linear regression. Let's view the new boxplots.

```python
fig, axes = plt.subplots(3, 3, figsize=(10,10))
fig.tight_layout(pad=3)

sns.boxplot(x=movies['Runtime'], ax=axes[0,0], color='skyblue');
sns.boxplot(x=movies['Budget'], ax=axes[0,1], color='skyblue');
sns.boxplot(x=movies['Revenue'], ax=axes[0,2], color='skyblue');

sns.boxplot(x=movies['IMDb'], ax=axes[1,0], color='lightgreen');
sns.boxplot(x=movies['Metacritic'], ax=axes[1,1], color='lightgreen');
sns.boxplot(x=movies['Rotten_Tomatoes'], ax=axes[1,2], color='lightgreen');

sns.boxplot(x=movies['Filmaffinity'], ax=axes[2,0], color='salmon');
sns.boxplot(x=movies['Personal'], ax=axes[2,1], color='salmon');
sns.boxplot(x=movies['Fans'], ax=axes[2,2], color='salmon');
```
