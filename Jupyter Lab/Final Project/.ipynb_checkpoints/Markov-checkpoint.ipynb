```python
import numpy as np

# Define the states
States = ["Jabon para la cara", "Crema hidratante", "Serum"]
NDays = 140


TransitionMatrix = np.array([[0.3, 0.44, 0.26],
[0.44, 0.3, 0.26],
[0.26, 0.44, 0.3]])


InitialState = np.array([0.44, 0.37, 0.19]) # Starting in state 0

State = InitialState
for day in range(1, NDays + 1):
    NewState = np.dot(State, TransitionMatrix)
    State = NewState
    MostLikelyState = np.argmax(State)
    print(f"Day {day}: Most likely state is {States[MostLikelyState]} with probabilities {State}")

print(State)
```
