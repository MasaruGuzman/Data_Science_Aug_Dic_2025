#### Detecting outliers

In order to detect outliers, we generated boxplots for all numerical columns in the dataset.

Purpose:
- Identify outliers visually across all key metrics.
- Boxplots highlight data spread, median, and extreme values.
- Helps detect unusual records that may require further attention, transformation, or exclusion.

```python
fig, axes = plt.subplots(3, 3,figsize=(10,10))
fig.tight_layout(pad=3)
sns.boxplot(x=movies['Runtime'], ax=axes[0,0], color='skyblue');
sns.boxplot(x=movies['Budget'], ax=axes[0,1], color='skyblue');
sns.boxplot(x=movies['Revenue'], ax=axes[0,2], color='skyblue');

sns.boxplot(y=movies['IMDb'], ax=axes[1,0], color='lightgreen');
sns.boxplot(y=movies['Metacritic'], ax=axes[1,1], color='lightgreen');
sns.boxplot(y=movies['Rotten_Tomatoes'], ax=axes[1,2], color='lightgreen');

sns.boxplot(x=movies['Filmaffinity'], ax=axes[2,0], color = "salmon");
sns.boxplot(x=movies['Personal'], ax=axes[2,1], color = "salmon");
sns.boxplot(x=movies['Fans'], ax=axes[2,2], color = "salmon");
```

Our approach for removing outliers will be naive: We calculate the interquartile range, `IQR`, for all numerical variables, and remove all records more than `1.5*IQR` units from the nearest quartile. This reduces the interquartile range and will possible "create" more outliers, but they will be within the endpoints of the whiskers of the previous plots.

```python
movies_numeric = movies.select_dtypes(include='number')
```

- `Q1`: 25th percentile (lower quartile).
- `Q3`: 75th percentile (upper quartile).
- `IQR = Q3 - Q1`: the range that contains the middle 50% of the data.

```python
Q1 = movies_numeric.quantile(0.25)
Q3 = movies_numeric.quantile(0.75)
IQR = Q3 - Q1
IQR
```

We detect outliers by filtering rows where **any numeric value** falls outside the acceptable IQR range:
Purpose:
- Detect movies with unusual values (extremely high budget, runtime, ratings, etc.).

```python
movies_outliers=movies[((movies_numeric < (Q1 - 1.5 * IQR)) |(movies_numeric > (Q3 + 1.5 * IQR))).any(axis=1)]
movies_outliers
```

We remove all rows identified as outliers using their index from the `movies_outliers` DataFrame.

```python
movies.drop(movies_outliers.index,inplace=True)
movies.shape
```

We are left with 166 records, still enough for linear regression. Let's view the new boxplots.

```python
fig, axes = plt.subplots(3, 3, figsize=(10,10))
fig.tight_layout(pad=3)

sns.boxplot(x=movies['Runtime'], ax=axes[0,0], color='skyblue');
sns.boxplot(x=movies['Budget'], ax=axes[0,1], color='skyblue');
sns.boxplot(x=movies['Revenue'], ax=axes[0,2], color='skyblue');

sns.boxplot(y=movies['IMDb'], ax=axes[1,0], color='lightgreen');
sns.boxplot(y=movies['Metacritic'], ax=axes[1,1], color='lightgreen');
sns.boxplot(y=movies['Rotten_Tomatoes'], ax=axes[1,2], color='lightgreen');

sns.boxplot(x=movies['Filmaffinity'], ax=axes[2,0], color='salmon');
sns.boxplot(x=movies['Personal'], ax=axes[2,1], color='salmon');
sns.boxplot(x=movies['Fans'], ax=axes[2,2], color='salmon');
```

#### Correlations

Now we create some plots that display the correlation between variables.

We use `sns.pairplot()` to generate a matrix of plots showing pairwise relationships between all numerical variables in the dataset.

Parameters used:
- `kind='reg'`: includes linear regression lines in the scatter plots to highlight trends.
- `diag_kind='kde'`: uses kernel density estimates on the diagonals to show variable distributions.
- `height=2.5`: sets the size of each subplot for better readability.

Purpose:
- Explore correlations between variables (e.g. budget vs revenue, IMDb vs Metacritic).
- Identify linear relationships or clusters.
- Spot potential multicollinearity or variable redundancy.

```python
sns.pairplot(movies, height=2.5, kind = 'reg', diag_kind='kde')
```

### Interpreting the pairplot

The pairplot shows pairwise relationships between all numerical variables in the dataset.

**1. Distributions (diagonal plots)**  
- `Budget` and `Revenue` are clearly right-skewed: most movies have relatively low values, with a few very expensive / very profitable titles.  
- The different rating variables (`Personal`, `IMDb`, `Rotten_Tomatoes`, `Metacritic`, `Filmaffinity`, `Fans`) are concentrated in the upper part of their scales, with few very low scores.  
- `Runtime` is concentrated in a relatively narrow band, with no extreme outliers in duration.

**2. Relationships between ratings**  
- All rating variables are positively correlated: when one score is high, the others tend to be high as well.  
- Some pairs show an almost straight line with a clear upward trend, indicating a strong linear relationship between those two rating sources (they “agree” a lot on which movies are good or bad).  
- This suggests that different rating systems and the `Fans` / `Personal` scores are broadly consistent with each other.

**3. Budget and revenue**  
- There is a positive relationship between `Budget` and `Revenue`: higher-budget movies tend to earn more money.  
- The scatter is wide, meaning budget alone does **not** guarantee high revenue; there are both successful and unsuccessful high-budget movies.

**4. Ratings vs money / runtime**  
- The relationship between ratings (IMDb, critics, fans, etc.) and `Budget` or `Revenue` is much weaker: points form a broad cloud with only a mild trend at best.  
  > Expensive or high-grossing movies are not always the best rated.  
- `Runtime` shows weak or no clear correlation with most other variables: longer movies are not systematically better rated or more profitable.

**5. Outliers**  
- Some isolated points in `Budget` and `Revenue` stand far from the bulk of the data, confirming the presence of extreme values already seen in the boxplots.  
- These outliers can strongly influence correlations and should be kept in mind (or removed, as done earlier) for modeling.

**Summary:**  
Ratings are strongly correlated with each other, budget is loosely correlated with revenue, and most other relationships are weak. The dataset contains a few extreme movies in terms of budget/revenue that behave as outliers.
