## Descriptive statistics
### Minimum and maximum

Minimum and maximum can be found thanks to the `pd.min()` and `pd.max()` methods

##### IMDb ratings

```python
imdb.min(), imdb.max()
 # Type the code to obtain the minimum and maximum of IMDb ratings here
```

- *What is the minimum IMDb rating?* <u>31</u>
- *What is the maximum IMDb rating?* <u>93</u>

#### Personal ratings

```python
personal.min(), personal.max() # Type the code to obtain the minimum and maximum of your personal ratings here
```

- *What is the minimum personal rating?* <u>20</u>
- *What is the maximum personal rating?* <u>100</u>

### Measures of central tendency
#### Mean
The mean can be computed with the `pd.mean()` method:

```python
imdb.mean(), personal.mean() # Type the code for obtaining the mean of the extracted variables
```

- *What is the average IMDb rating?* <u>74.05</u>
- *What is your average personal rating?* <u>81.457317</u>

#### Median

The median can be computed thanks to the `pd.median()` method:

```python
imdb.median(), personal.median() # Type the code for obtaining the median of the extracted variables
```

- *What is the median of IMDb ratings?* <u>75.0</u>
- *What is the median of your personal ratings?* <u>80.0</u>

### Measures of variation
#### Range
The range of the data can be easily computed by subtracting the minimum from the maximum

```python
imdb.max()-imdb.min(), personal.max()-personal.min() # Type the code for obtaining the range of both variables here
```

- *What is the range of IMDb ratings?* <u>62</u>
- *What is the range of your personal ratings?* <u>80.0</u>

#### Interquartile range
In order to calculate the interquartile range (i.e., the difference between the first and third quartiles), we first extract said quartiles with the `pd.quantile()` method and compute their difference

```python
imdb.quantile(), personal.quantile() # Type the code for obtaining the interquartile range of both variables here
```

- *What is the interquartile range of IMDb ratings?* <u>75.0</u>
- *What is the interquartile range of your personal ratings?* <u>80.0</u>

#### Standard deviation and variance
The standard deviation and the variance are computed with the `pd.std()` and `pd.var()` methods, respectively. These methods correspond to the sample standard deviation and variance, since Python assumes that the data corresponds to a sample of the population.

```python
imdb.var(), personal.var(), imdb.std(), personal.std() # Type the code for obtaining the variance and standard deviation of both variables here
```

- *What is the variance of IMDb ratings?* <u>85.645729</u>
- *What is the variance of your personal ratings?* <u>138.114731</u>
- *What is the standard deviation of IMDb ratings?* <u>9.254498</u>
- *What is the standard deviation of your personal ratings?* <u>11.752222</u>

### Histogram

A histogram gives an idea about the distribution of a quantitative variable. The idea is to break the range of values into intervals and count how many observations fall into each interval. Histograms are a bit similar to barplots, but histograms are used for quantitative variables whereas barplots are used for qualitative variables.

```python
imdb_histogram = sns.histplot(imdb, bins = 10, kde = False, legend = False);
imdb_histogram.set(title = 'IMDb Histogram', xlabel = 'Score', ylabel = 'Movies');
plt.show()# Type the code for drawing a histogram for IMDb ratings with 10 bins here
```

- *Does the skewness of the histogram agree with the skewness parameter given in the description of the data above?* <u>Yes</u>

```python
personal_histogram = sns.histplot(personal, bins = 10, kde = False, legend = False);
personal_histogram.set(title = 'Personal Histogram', xlabel = 'Score', ylabel = 'Movies');
plt.show()# Type the code for drawing a histogram for your personal ratings with 10 bins here
```

- *Does the skewness of the histogram agree with the skewness parameter given in the description of the data above?* <u>Yes</u>

### Boxplots

Boxplots are really useful in descriptive statistics and are often underused (mostly because it is not well understood by the public). A boxplot graphically represents the distribution of a quantitative variable by visually displaying five common location summary (minimum, median, first/third quartiles and maximum) and any observation that was classified as a suspected outlier using the interquartile range (IQR) criterion.

The IQR criterion means that all observations above $q_{0.75}+1.5\cdot IQR$ or below $q_{0.25}-1.5\cdot IQR$ (where $q_{0.25}$ and $q_{0.75}$ correspond to first and third quartile respectively) are considered as potential outliers by R. The minimum and maximum in the boxplot are represented without these suspected outliers.

Seeing all these information on the same plot help to have a good first overview of the dispersion and the location of the data.

```python
fig, ax = plt.subplots()
sns.boxplot(imdb, color = 'LightBlue', linecolor = 'blue', orient = 'h')
ax.set(title = 'IMDb Boxplot', xlabel = 'Score')
plt.show()# Type the code for drawing a boxplot for IMDb ratings 
```

- *Are there any outliers for this variable?* <u>Yes, there are 2</u>

```python
fig, ax = plt.subplots()
sns.boxplot(personal, color = 'LightBlue', linecolor = 'blue', orient = 'h')
ax.set(title = 'Personal Boxplot', xlabel = 'Score')
plt.show()# Type the code for drawing a boxplot for Personal ratings 
```

- *Are there any outliers for this variable?* <u>Yes, there are 4</u>
